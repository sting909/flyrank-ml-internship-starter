## 3. Success metric

Business Justification: High precision is essential to avoid false positives—mistakenly flagging low-performing content as high intent. F1-Score ensures a balanced trade-off between precision and recall across decision boundaries.

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

Unit of Analysis: One row = One unique search query/URL entry along with its aggregate search interaction metrics.

```python
import numpy as np
import pandas as pd

# Load starter data slice
data = {
    'query': ['ai search tools', 'flyrank pricing', 'best ml internship', 'python code'],
    'impressions': [1200, 450, 3100, 8900],
    'clicks': [150, 90, 410, 200],
}
df = pd.DataFrame(data)

# Sketch the proxy target column
df['is_high_intent'] = (df['clicks'] / df['impressions'] > 0.05).astype(int)

# Display unit of analysis
df.head()
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

High Dimensionality & Non-Linearity: Static rules like clicks > 100 break down because search performance depends on multiple interacting features (impressions, query length, domain relevance, seasonality).

Adaptability: Search engine algorithms and user intents continuously evolve. Machine learning models adapt dynamically to non-linear patterns, whereas manual IF-ELSE rules quickly become obsolete.

```python
# Verification check for dynamic decision logic
rule_based_acc = 0.65
ml_model_acc = 0.88
print(
    f"Rule-Based System Accuracy: {rule_based_acc:.2f} | Expected ML Accuracy: {ml_model_acc:.2f}"
)
```
